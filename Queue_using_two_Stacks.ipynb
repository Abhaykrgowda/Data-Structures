2. Queue Using two Stacks

```python
class MyQueue:
  def __init__(self):
    self.stack_in = []
    self.stack_out = []

  def enqueue(self, value):
    self.stack_in.append(value)

  def dequeue(self):
    self._move()
    return self.stack_out.pop() if self.stack_out else None

  def peek(self):
    self._move()
    return self.stack_out[-1] if self.stack_out else None

  def _move(self):
    if not self.stack_out:
      while self.stack_in:
        self.stack_out.append(self.stack_in.pop())

```

```python
queue = MyQueue()
queue.enqueue(1)
queue.enqueue(4)
queue.enqueue(9)
print(queue.peek())
print(queue.dequeue())
print(queue.peek())
```
