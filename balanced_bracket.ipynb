1. Balanced Brackets

```python
def is_balanced(s):
  stack = []
  bracket_maping = {')':'(','}':'{',']':'['}
  for char in s:
    if char in ['(','{','[']:
      stack.append(char)

    elif char in [')','}',']']:
      if stack == [] or bracket_maping[char]!=stack.pop():
        return False
  return True if not stack else False
```

```python
s = '[{(()}]'
print(is_balanced(s))
```
